```python
import numpy as np
import pandas as pd

def etl(telemetry):
    reading_c = np.array(telemetry["reading_c"])
    baseline_c = np.array(telemetry["baseline_c"])

    delta_c = np.round(reading_c - baseline_c, 1)
    status = np.where(np.abs(delta_c) >= 10, "alert", "normal")

    result = pd.DataFrame({
        "device_id": telemetry["device_id"],
        "delta_c": delta_c,
        "status": status
    })

    return result.sort_values("device_id").reset_index(drop=True)


telemetry = pd.DataFrame({
    "device_id": ["d3", "d1", "d2"],
    "reading_c": [65, 72.5, 95],
    "baseline_c": [70, 70, 80]
})

result = etl(telemetry)
print(result)
```
