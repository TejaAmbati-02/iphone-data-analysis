```python
df.groupby('Model Name')['Mrp'].sum()


# How it works:
# - groupby('Model Name') groups rows by each model.
# - ['Mrp'] selects the MRP column.
# - .sum() calculates the total MRP for each model.
```

```python
# Total MRP per model
df.groupby('Model Name')['Mrp'].sum()

# Average MRP per model
df.groupby('Model Name')['Mrp'].mean()

# Number of MRP entries per model
df.groupby('Model Name')['Mrp'].count()

# Multiple statistics per model
df.groupby('Model Name')['Mrp'].agg(['sum', 'mean', 'count'])
```
