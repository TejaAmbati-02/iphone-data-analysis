```python
import pandas as pd

def etl(orders):
    orders = orders[orders['status'].str.lower() == 'completed'].copy()

    orders[['amount', 'discount_rate']] = orders[
        ['amount', 'discount_rate']
    ].fillna(0)

    orders['net_amount'] = (
        orders['amount'] * (1 - orders['discount_rate'])
    ).round(2)

    return (
        orders[['order_id', 'channel', 'net_amount']]
        .sort_values('order_id')
        .reset_index(drop=True)
    )
```

```python
import pandas as pd

orders = pd.DataFrame({
    'order_id': [103, 101, 102, 104],
    'channel': ['Web', 'App', 'Web', 'Store'],
    'status': ['completed', 'Completed', 'pending', 'completed'],
    'amount': [200, 100, 150, None],
    'discount_rate': [0.1, None, 0.2, 0.15]
})

result = etl(orders)
print(result)
```

```python
# Join
left_df = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2']
}, index=['K0', 'K1', 'K2']) 

right_df = pd.DataFrame({
    'C': ['C0', 'C2', 'C3'],
    'D': ['D0', 'D2', 'D3']
}, index=['K0', 'K2', 'K3'])
```

```python
left_df.join(right_df, how='inner')
```

```python
# Merging on multiple keys
left = pd.DataFrame({
    'key1': ['K0', 'K0', 'K1', 'K2'],
    'key2': ['K0', 'K1', 'K0', 'K1'],
    'A': ['A0', 'A1', 'A2', 'A3'],
    'B': ['B0', 'B1', 'B2', 'B3']
})
    
right = pd.DataFrame({
    'key1': ['K0', 'K1', 'K1', 'K2'],
    'key2': ['K0', 'K0', 'K0', 'K0'],
    'C': ['C0', 'C1', 'C2', 'C3'],
    'D': ['D0', 'D1', 'D2', 'D3']
})
```

```python
left
```

```python
right
```

```python
# other options are 'inner', 'left', 'right'
pd.merge(left, right, how='left', on=['key1', 'key2'])
```
